**Multiple Exception**

```python
try:
  b= int(input("Enter numerical value :"))
  print(10/b)

except ZeroDivisionError:
  print("Divided by zero.")

except ValueError:
  print("ENter only numeric value.")

finally:
  print("I always run hahaha")
```

**Custom Exception**

```python
class SectionA(Exception):
    pass

class SectionB(Exception):
  pass

class SectionC(Exception):
  pass


marks=90

try:
  if marks>=80:
    raise SectionA("You go to section A")

  elif marks<=33:
    raise SectionC("You go to Section C")

  elif marks>=34 & marks<=79:
    raise SectionB("You go to SectionB")

except SectionA as e:
  print (e)

except SectionB as e:
  print(e)

except SectionC as e:
  print(e)
```
