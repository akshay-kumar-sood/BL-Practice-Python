```python
class Student:
  ''' Add 2 objects example of method overloading'''

  # Constructor
  def __init__(self,m1,m2):
    self.m1=m1
    self.m2=m2

 # This works when we add 2 objects using + operator
 # + operator equivalent to __add__

  def __add__(self,other):
    m1=self.m1 + other.m1
    m2=self.m2 + other.m2
    s3=Student(m1,m2)
    return s3


s1=Student(10,20)
s2=Student(40,50)

s3=s1+s2
print(s3.m1)
print("\n")
print(s3.m2)
```

```python
class Salary:

  ''' we have to add salary of an employee of 2 month '''

  def __init__(self,m1):
    self.m1=m1

  def __add__(first,second):
    m1=first.m1 + second.m1
    total=Salary(m1)
    return total

  # Differenece between there salary

  def __sub__(first,second):
    m1=first.m1-second.m1
    total=Salary(m1)
    return total

jan=Salary(3000)
feb=Salary(2000)

total=jan+feb
print(total.m1)

diff=jan - feb
print(diff.m1)
```
