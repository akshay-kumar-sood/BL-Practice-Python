# **Regular Expression:**

```python
import re

quote = "I scream, you scream, we all scream for ice cream."

search_method = re.search("dream",quote)
print(f"Output by using Search method: {search_method}")

search_method = re.search("scream",quote)
print(f"Output by using Search method: {search_method}")
```

**Explanation:**
* The `re.search()` method scans through the string to find the first location where the regex pattern matches.
* In this case, you're searching for the word "dream" in the quote.
* Since "dream" does not exist in the quote, re.search() will return None.


```python

findall_method = re.findall("scream", quote)
print(f"Output by using Findall method: {findall_method}")

findall_method = re.findall("dream", quote)
print(f"Output by using Findall method: {findall_method}")
```

**Explanation:**

* The `re.findall()` method returns all non-overlapping matches of the pattern in the string as a list.
* In this case, you're searching for the word "scream" in the quote.
* The word "scream" appears 3 times in the quote, so re.findall() will return a list containing each occurrence.

```python
split_method = re.split(",",quote)
print(f"Output by using Split method: {split_method}")
```

**Explanation:**

* The `re.split()` method splits the string by the occurrences of the pattern and returns a list of the resulting substrings.
* In this case, you're splitting the quote by the comma ,.
* The quote contains two commas, so re.split() will divide the string into three parts.

# 1. \d (Matches any decimal digit [0-9])

```python
# \d (Matches any decimal digit [0-9])

# Example 1
text1 = "I have 2 apples and 5 oranges."

matches1 = re.findall(r'\d+', text1)
print(f"Output of \d: {matches1}")

# Example 2
text2 = "The year is 2024."

matches2 = re.findall(r'\d{2}', text2)
print(f"Output of \d{{2}}: {matches2}")
```

**Explanation:** \

**Example 1:** \

* **r'\d+':** The \d matches any digit (0-9), and the + means one or more occurrences. So, this regex finds one or more digits in a row. \
* **re.findall():** This function returns all occurrences of the pattern in the string. \
* **matches1:** Will contain `['2', '5']` because those are the digit sequences in the text. \

**Example 2:** \
* **r'\d{2}':** The \d matches any digit, and {2} specifies exactly two occurrences. \
* **matches2:** Will contain `['20', '24']` because it's a two-digit sequence in the text.

# 2. \D (Matches any non-digit character [^0-9])

```python
#  \D (Matches any non-digit character [^0-9])

# Example 1
text1 = "123 Main St. Francis"
matches1 = re.findall(r'\D+', text1)
print(f"Output of \D+: {matches1}")

# Example 2
text2 = "House number: 45A"
matches2 = re.findall(r'\D{3}', text2)
print(f"Output of \D{{3}}: {matches2}")
```

**Explanation:** \

**Example 1:** \

* **text1 = "123 Main St. Francis":** The string contains both digits and non-digit characters.
* **r'\D+':**
\D matches any character that is not a digit.
* **\+** is a quantifier that means "one or more" occurrences of the preceding pattern.\
So, \D+ matches sequences of one or more non-digit characters.
* **re.findall(r'\D+', text1):**
re.findall() finds all non-overlapping matches of the pattern in the string and returns them as a list.\
In this case, it will return sequences of non-digit characters from text1.

**Example 2:** \
* **text2 = "House number:** 45A": The string contains both digits and non-digit characters.
* **r'\D{3}':**
\D matches any character that is not a digit.
{3} is a quantifier that specifies "exactly three" occurrences of the preceding pattern. \
So, \D{3} matches sequences of exactly three non-digit characters.
* **re.findall(r'\D{3}', text2):**
re.findall() finds all non-overlapping matches of the pattern in the string and returns them as a list. \
In this case, it will return sequences of exactly three non-digit characters from text2.



# 3. \s (Matches any whitespace character [ \t\n\r\f\v])

**Explanation:**

**Example 1:** \

* **text1 = "Hello World !":** The string contains words separated by spaces.
* **r'\s+':**
\s matches any whitespace character.
* **+** is a quantifier that means "one or more" occurrences of the preceding pattern.\
So, \s+ matches sequences of one or more whitespace characters.
* **re.findall(r'\s+', text1):**
re.findall() searches for all non-overlapping matches of the pattern in the string and returns them as a list.\
In this case, it will return sequences of whitespace characters from text1.

**Example 2:** \

* **text2 = "Python** is awesome.": The string contains words with spaces, including a sequence of three consecutive spaces.
* **r'\s{2}':**
\s matches any whitespace character.
{2} is a quantifier that specifies "exactly two" occurrences of the preceding pattern.\
So, \s{2} matches sequences of exactly two whitespace characters.
* **re.findall(r'\s{2}', text2):**
re.findall() finds all non-overlapping matches of the pattern in the string and returns them as a list.\
In this case, it will return sequences of exactly two consecutive whitespace characters from text2.

# 5. \w (Matches any alphanumeric character [a-zA-Z0-9_])

**Explanation:** \

* **re.findall(r'\w+', text1):**
re.findall() searches for all non-overlapping matches of the pattern in the string and returns them as a list. \
In this case, it will return sequences of alphanumeric characters and underscores from text1.

**Example 2:** \

* **text2 = "A1_B2":** The string contains sequences of alphanumeric characters separated by underscores.
* **r'\w{2}':**
\w matches any alphanumeric character or an underscore `([a-zA-Z0-9_])`.
* `{2}` is a quantifier that specifies "exactly two" occurrences of the preceding pattern. \
So, \w{2} matches sequences of exactly two alphanumeric characters or underscores.
* **re.findall(r'\w{2}', text2):**
re.findall() finds all non-overlapping matches of the pattern in the string and returns them as a list. \

In this case, it will return sequences of non-alphanumeric characters from text1.

**Example 2:** \

* **text2 = "@Good!?#":** The string contains a mix of alphanumeric and non-alphanumeric characters.
* **r'\W{3}':**
\W matches any character that is not alphanumeric `([^a-zA-Z0-9_])`.
* `{3}` is a quantifier that specifies "exactly three" occurrences of the preceding pattern.\
So, \W{3} matches sequences of exactly three non-alphanumeric characters.
* **re.findall(r'\W{3}', text2):**
